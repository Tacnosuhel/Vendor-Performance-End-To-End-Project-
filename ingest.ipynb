```python
import logging
import time
import pandas as pd
import os 
logging.basicConfig(
  filename="logs/ingestion_db.log",
  level=logging.DEBUG,
  format="%(asctime)s - %(levelname)s - %(message)s",
  filemode="a"
)
```

```python
# Data ingestion
def ingest_db(file, table_name, engine):

    for chunk in pd.read_csv(file, chunksize=5000):
        chunk.to_sql(
            table_name,
            con=engine,
            if_exists="append",
            index=False
        )
```

```python
def load_row_data():

    start = time.time()

    for file in os.listdir("data"):

        if file.endswith(".csv"):

            logging.info(f"Ingestion started: {file}")

            ingest_db(
                f"data/{file}",
                file[:-4],
                engine
            )

    total_time = (time.time() - start) / 60

    logging.info("----------Ingestion Completed------------")
    logging.info(f"Total Time: {total_time:.2f} minutes")
```

```python
# Function Load 
load_row_data()
```
